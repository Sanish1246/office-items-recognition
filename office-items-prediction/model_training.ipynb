## Importing Dependencies

```python
# imports
from ultralytics import YOLO
from pathlib import Path
```

## Define Paths & folder structure

```python
# data path
DATA_DIR = Path("data/split")

# verify folder structure
assert (DATA_DIR / "train").exists(), "Train folder missing!"
assert (DATA_DIR / "val").exists(), "Validation folder missing!"
assert (DATA_DIR / "test").exists(), "Test folder missing!"
```

## Set up of Yolo dataset config

```python
yaml_path = Path("office_config.yaml")
if not yaml_path.exists():
    CLASS_NAMES = sorted([p.name for p in (DATA_DIR / "train").iterdir() if p.is_dir()])
    yaml_content = f"""
# YOLOv8 Classification Dataset Config
path: {DATA_DIR}
train: train
val: val
test: test

names:
""" + "\n".join([f"  {i}: {cls}" for i, cls in enumerate(CLASS_NAMES)])
    yaml_path.write_text(yaml_content.strip())
    print(f"{yaml_path} was created successfully")
```

## Training Model

```python
# using yolo medium model
model = YOLO("yolov8m-cls.pt")

# train
results = model.train(
    data=str("data/split"),
    epochs=15,
    imgsz=224,     
    batch=32,
    lr0=0.001, 
    project="runs/office_cls",
    name="trial_run_1",
    verbose=True
)
```
