```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11s-cls.pt")  # load a pretrained runs (recommended for training)
```

```python
results = model.train(data="data\\Image Classification", epochs=150, imgsz=320, batch=-1)
```

```python
# Validate the runs
model = YOLO(r"models\train\weights\best.pt")
metrics = model.val(data=r"data\Image Classification")  # no arguments needed, dataset and settings remembered
print(metrics.top1)  # top1 accuracy
print(metrics.top5)  # top5 accuracy
```

```python
# Load a runs
model = YOLO(r"models\train\weights\best.pt")  # load a custom runs

# Predict with the runs
results = model("https://hips.hearstapps.com/hmg-prod/images/ferrari-e-suv-2-copy-680287cac36b2.jpg?crop=1.00xw:0.838xh;0,0.0673xh")  # predict on an image

# Process results list
for result in results:
    probs = result.probs  # Probs object for classification outputs
    result.show()  # display to screen
    result.save(filename=r"prediction\result.jpg")  # save to disk
```
